### Leap Year Logic Write a function that takes a year as input and returns True if it is a leap year and False otherwise. 

```python
year = int(input("Enter a year: "))

def isLeap(year):
    isLeap=False
    if year%400==0:
        isLeap=True
    elif year%100!=0 and year%4==0:
        isLeap=True
    return isLeap

print("Is leap : ", isLeap(year))
```

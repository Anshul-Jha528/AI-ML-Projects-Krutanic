###  Identity vs Equality Given: L1 = [1,2,3] and L2 = [1,2,3] Explain the difference between L1 == L2 and L1 is L2

```python
l1 = [1,2,3]
l2 = [1,2,3]

print("L1==L2: ",(l1==l2))
print("L1 is L2: ",(l1 is l2))
```

#### L1==L2 compares elements of the list and return true if all elements are equal, while L1 is L2 com[ares the memory location of object, returns true only if they are same object

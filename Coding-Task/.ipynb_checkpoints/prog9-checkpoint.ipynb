### Runner-Up Score Given a list of scores, find the second-highest score without using sort() or sorted(). 

```python
scores = [60,85,99,87,43,55,66,80]

high = -1
sec_high = -1

for score in scores:
    if score>high:
        sec_high = high
        high = score
    elif score>sec_high:
        sec_high = score

print("Second highest score : ",sec_high)
```

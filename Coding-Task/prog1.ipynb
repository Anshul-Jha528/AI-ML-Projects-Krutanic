###  FizzBuzz Challenge Write a program that iterates from 1 to 50. - Print 'Fizz' for multiples of 3 - Print 'Buzz' for multiples of 5 - Print 'FizzBuzz' for multiples of both

```python
# Multiple of 3
for i in range(1,51):
    if i%3==0:
        print("Buzz")
```

```python
# Multiple of 5
for i in range(1,51):
    if i%5==0:
        print("Buzz")
```

```python
# Multiple of both
for i in range(1,51):
    if i%3==0 and i%5==0:
        print("Buzz")
```

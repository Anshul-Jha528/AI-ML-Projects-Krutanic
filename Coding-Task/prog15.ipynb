### Dictionary Comprehension From the dictionary {'a': 1, 'b': 2, 'c': 3, 'd': 4}, create a new dictionary containing only items with values greater than 2

```python
dict1 = {'a': 1, 'b': 2, 'c': 3, 'd': 4}

dict2 = {}

for  key in dict1:
    if dict1[key]>2:
        dict2[key]=dict1[key]

print(dict2)
```

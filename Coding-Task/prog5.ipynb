###  Prime Number Finder- Write a while loop that finds and prints the first 10 prime numbers. 

```python
def isPrime(x):
    for i in range(2,x):
        if x%i==0:
            return False
    return True
```

```python
i=0
x=2

while i<10:
    while not isPrime(x):
        x=x+1
    print(x)
    x=x+1
    i=i+1
```

###  Nested Conditionals Write a program that takes a numerical grade (0–100) and returns: A (90+), B (80–89), C (70–79), Fail (below 70).  

```python
marks = int(input("Enter your marks out of 100: "))

grade = ''

if marks>=90:
    grade='A'
elif marks>=80:
    grade='B'
elif marks>=70:
    grade='C'
else:
    grade='Fail'

print("Grade: ",grade)
```
